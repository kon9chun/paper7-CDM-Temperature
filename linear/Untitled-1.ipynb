```python
import numpy as np
true=np.load('val_ydata.npy')
pred=np.load('val_sp9_reconstructed.npy')
def calculate_rmse(matrix1, matrix2):
    error = matrix1 - matrix2    
    # 计算均方误差 (MSE)
    mse = np.mean(error ** 2)    
    # 计算均方根误差 (RMSE)
    rmse = np.sqrt(mse)    
    return rmse
linear_rmse=[]
for i in range(true.shape[0]):
    rmse=calculate_rmse(true[i],pred[i])
    linear_rmse.append(rmse)


np.save('linear_rmse.npy', np.array(linear_rmse))
np.array(linear_rmse).shape
```
